## Normalisasi customer review

Gunakan normalisasi ringan yang sama untuk TF-IDF, Gemini, dan model Hugging Face agar perbandingan adil. Kolom asli tetap disimpan. Tahap ini hanya menormalkan Unicode, mengganti karakter encoding yang rusak dengan spasi, dan merapikan whitespace; kapitalisasi, negasi, slang, tanda baca, dan emoji dipertahankan.

```python
import re
import unicodedata

def normalize_review(text):
    if pd.isna(text):
        return text

    text = unicodedata.normalize("NFKC", str(text))
    text = text.replace("\uFFFD", " ")
    return re.sub(r"\s+", " ", text).strip()

df_clean["customer_review_normalized"] = df_clean["customer_review"].map(normalize_review)

df_clean[["customer_review", "customer_review_normalized"]].head()
```

```python
df_clean["location"] = (
    df_clean["location"]
    .str.replace("\uFFFD", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
    .str.replace(r"(?i)^kab(?:upaten)?\.?\s+", "Kabupaten ", regex=True)
)

df_clean["location"].value_counts()
```

# Dataset Final

```python
df_clean["review_clean"] = df_clean["customer_review_normalized"]
```

```python
df_final = df_clean.copy()
log = {"awal": len(df_final)}

# 1. Buang baris tanpa teks atau label
df_final = df_final.dropna(subset=["customer_review", "sentiment"])
df_final = df_final[df_final["review_clean"].str.len() > 0]
log["setelah buang kosong"] = len(df_final)

# 2. Teks sama tapi label beda: buang semuanya (labelnya ambigu)
n_label = df_final.groupby("review_clean")["sentiment"].transform("nunique")
df_final = df_final[n_label == 1]
log["setelah buang teks berlabel ganda"] = len(df_final)

# 3. Buang teks duplikat, simpan yang pertama
df_final = df_final.drop_duplicates(subset="review_clean", keep="first")
log["setelah buang duplikat"] = len(df_final)

pd.Series(log)
```
